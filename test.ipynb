```python
import cv2

file_path = 'finals_verification/mp4_vids/phase1-rifle/cam3_alex_2.mp4'

cap = cv2.VideoCapture(file_path)

if not cap.isOpened():
    print("Error: Could not open video.")
    exit()

while True:
    ret, frame = cap.read()
    if not ret:
        break  # End of video

    cv2.imshow("Video", frame)

    # Press 'q' to exit early
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
